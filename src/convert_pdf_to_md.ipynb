```python

import pymupdf4llm
from pathlib import Path
```

```python
def convert_pdf_to_md(pdf_path: Path, output_path: Path) -> bool:
    """Converte un singolo PDF in Markdown. Ritorna True se successo."""
    try:
        md_text = pymupdf4llm.to_markdown(str(pdf_path))
        output_path.parent.mkdir(parents=True, exist_ok=True)
        output_path.write_text(md_text, encoding="utf-8")
        return True
    except Exception as e:
        print(f"  ⚠️  Errore su {pdf_path.name}: {e}")
        return False
```

```python
input_dir: Path = Path("../Dataset/knowledge_base")
output_dir: Path = Path("../Dataset/knowledge_base_md")
```

```python
pdf_files = sorted(input_dir.rglob("*.pdf"))
print(f"📂 Input:  {input_dir.resolve()}")
print(f"📂 Output: {output_dir.resolve()}")
print(f"📄 PDF trovati: {len(pdf_files)}\n")
```

```python
output_dir.mkdir(exist_ok=True)
```

```python
success, failed = 0, 0
 
for pdf_path in pdf_files:
    # Calcola path relativo e costruisci path di output
    relative = pdf_path.relative_to(input_dir)
    output_path = output_dir / relative.with_suffix(".md")

    print(f"  🔄 {relative}", end=" ... ")
    if convert_pdf_to_md(pdf_path, output_path):
        size_kb = output_path.stat().st_size / 1024
        print(f"✅  ({size_kb:.1f} KB)")
        success += 1
    else:
        failed += 1

print(f"\n{'='*50}")
print(f"✅ Convertiti con successo: {success}/{len(pdf_files)}")
if failed:
    print(f"❌ Falliti:               {failed}/{len(pdf_files)}")
print(f"📂 Output salvato in:      {output_dir.resolve()}")
print(f"{'='*50}")
```
